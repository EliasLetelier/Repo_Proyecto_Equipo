```python
import pandas as pd
import altair as alt
from google.colab import files
import io


uploaded = files.upload()
df = pd.read_csv(io.BytesIO(uploaded[list(uploaded.keys())[0]]), skiprows=1)
df.columns = [c.strip() for c in df.columns]


df['Público'] = pd.to_numeric(df['Público'], errors='coerce').fillna(0)


estadio_stats = df.groupby('Estadio').agg(
    Cantidad_Partidos=('Estadio', 'count'),
    Promedio_Publico=('Público', 'mean')
).reset_index()
estadio_stats['Promedio_Publico'] = estadio_stats['Promedio_Publico'].round(0)


chart = alt.Chart(estadio_stats).mark_bar().encode(
    y=alt.Y('Estadio:N', sort='-x', title='Recinto Deportivo'),
    x=alt.X('Cantidad_Partidos:Q', title='Partidos Jugados'),
    color=alt.Color('Promedio_Publico:Q', scale=alt.Scale(scheme='greens'), title='Promedio Público'),
    tooltip=['Estadio', 'Cantidad_Partidos', 'Promedio_Publico']
).properties(title='Frecuencia y Asistencia por Estadio', width=600)

chart
```
