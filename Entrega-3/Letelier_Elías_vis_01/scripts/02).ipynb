```python
import pandas as pd
import altair as alt
from google.colab import files
import io
```

```python
filename = list(uploaded.keys())[0]
df = pd.read_csv(io.BytesIO(uploaded[filename]), skiprows=1)
```

```python
df.columns = [c.strip() for c in df.columns]
df['Árbitro'] = df['Árbitro'].str.strip()
df['Ganador'] = df['Ganador'].str.strip()
```

```python
top_arbitros = df['Árbitro'].value_counts().nlargest(12).index
df_filtered = df[df['Árbitro'].isin(top_arbitros)].copy()
```

```python
viz_data = df_filtered.groupby(['Árbitro', 'Ganador']).size().reset_index(name='Cantidad')
```

```python
chart = alt.Chart(viz_data).mark_bar().encode(
    y=alt.Y('Árbitro:N',
            title='Árbitro',
            sort='-x'),
    x=alt.X('Cantidad:Q',
            title='Número de Partidos'),
    color=alt.Color('Ganador:N',
                    scale=alt.Scale(
                        domain=['U de Chile', 'Colo-Colo', 'Empate'],
                        range=['#003399', '#000000', '#808080']
                    ),
                    title="Resultado Final"),
    order=alt.Order('Ganador:N', sort='ascending'),
    tooltip=['Árbitro', 'Ganador', 'Cantidad']
).properties(
    width=600,
    height=450,
    title='Historial por Árbitro: Victorias y Empates (Superclásico)'
).configure_title(
    fontSize=18,
    anchor='start'
)
```
